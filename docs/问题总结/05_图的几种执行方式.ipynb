## 图的几种执行方式
1. 同步/异步 的基本调用
.invoke(input,config) --同步，跑完整个图，只返回最终的状态，没有中间结果
.ainvoke(input,config)  -- invoke的异步版本

2. stream 系列 -- 流式输出中间结果
.stream(input,config,stream_mode=...) -- 同步生成器，每个节点执行完就yield一次状态更新
.astream(...) -- 异步版本

stream_mode 决定了yield的粒度
- values(默认)：每步yield完整的当前state
- updates：每步只yield这一步节点返回的增量
- massgages： 流式输出LLM生成的token(适合聊天打字机效果)
- debug: 详细调试内容

特点，只能看到节点级别的进展，看不到节点内部调用了哪些工具

3. astream_events(input,config,versin="v2")
异步生成器，事件粒度比astream细得多，能穿透节点内部，捕获：
- on_chain_start/on_chain_end(每个节点/子图)asd
- on_tool_start/on_tool_end(工具调用)
- on_llm_start/on-llm_stream/on_llm_end(LLM流式token)

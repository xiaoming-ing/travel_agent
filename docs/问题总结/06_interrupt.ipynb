## interrupt 
- 当某个节点调用interrupt的时候，LangGraph会暂停整个图的执行，并把"下一个要跑的节点"记录为一个待执行的task,连同interrupt()传入的那个dict一起存进checkpoint
- graph.aget_state(config)把这个checkpointe读取出来，state.tasks就会有这个待执行的feedback task,且task.interrupts 非空，里面的.value就是在interrupt({'type':'feedback'..})传的dict
- 反之，如果图已经走到END，已经没有下一个待执行的节点了，state.tasks是空元组()

恢复时候是从挂起的那个节点开始的，从函数角度看，是那个节点函数从头开始跑的，不是断点那行开始的

checkpointer 起到保存快照、支持断点续跑的作用

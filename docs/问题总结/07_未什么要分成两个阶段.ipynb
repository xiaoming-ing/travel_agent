## 分成两阶段的原因是 两件事的技术要求冲突，一个agent无法满足
第一阶段需要的是**自主决策能力**
查什么工具、查几次、顺序怎么排，这些取决于用户输入，没有固定答案。ReAct Agent最擅长这个--让LLM自己循环决策，直到数据够了为止。
但 ReAct Agent的输出是自由文本，结构不可控

第二阶段需要的是**严格的结构化输出**
TripPlan 有固定字段，类型、嵌套层级，前端直接消费，不能有任何格式偏差，with_structured_output能保证这点，但它只调用一次LLM，没有根据情况决定要不要再去查一次数据的 能力

为什么要分两个阶段呢
因为前端是结构化的展示页面，有景点、酒店、天气、地图模块，每个，消费的是固定字段，前端写死要这些字段，所以后端输出必须是固定的结构，可以说是前端倒逼出来的这个设计。
如果前端改成像DeepSearch那样的对话式的，那这个第二阶段是完全可以去掉的，就是按照一阶段的输出渲染markdown就行

### prompt的编写思路
- Role 你是谁
- Goal 最终要完成什么
- Inputs 你会收到什么信息
- Tools/Knowledge 你有哪些工具、RAG、数据源
- Rules 必须遵守什么规则
- Output / Behavior 最终怎么输出、什么时候追问、什么时候调用工具

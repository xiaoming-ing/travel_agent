### 遇到的问题

#### 总结节点会重复执行
```
builder = StateGraph(TravelState)
builder.add_edge("weather", "itinerary")
builder.add_edge("hotel", "itinerary")
builder.add_edge(["weather","hotel"], "itinerary")
```
weather_node指向总结节点，hotel_node也指向总结节点，导致每个节点执行完都会触发总结节点，langgraph设计就是如此  
解决方法：我首先想到的是在总结节点做数据判断，为齐全的话直接return  
后面我又看了问了下AI其实有更简单的方法，就是在图哪里把两个节点放到一个数组里面，这样的话就是确保两个节点都执行完才会执行最终的总结节点


#### 为什么需要后处理
LLM有几个天然问题
- 幻觉
- 数据不一定和真实数据一致
- 格式对了，但是内容不可信
